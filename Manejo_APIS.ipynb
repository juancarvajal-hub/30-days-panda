API: Application Programming Interface

Las API son mecanismos que permiten a dos componentes de software comunicarse entre sí mediante un conjunto de definiciones y protocolos.

Estudiaremos las API REST, transferencia de datos representacional, son muy usandas y usa sentencias HTTP.

Concepotos claves 
- Endpoint: URL donde está el recurso 
- Método: Acción que quieres hacer (GET,POST,PUT,DELETE)
- Recursos: Lo que pides (Personajes, Ususarios, Reursos)
- Payload: Datos que envias en (POST,PUT)
- Status Code: Número que indica el resultado (200 = ok, 404 = No encontrado)


```python
import tenacity # Para reintentos
import requests 
import pandas as pd

## Primera conexión 

# Una Api simple para proabar 

url = "https://jsonplaceholder.typicode.com/posts/1"

# Hacemos la petición 
response = requests.get(url)

# Ver el resultado
print(response.status_code)        # 200
print(response.text)               # Texto crudo, datos 
print(response.json())             # Como diccionario Python
```

 Que está sucediendo en los pasos anteriores:

 1. get hace una petición HTTP.
 2. El servidor responde con datos en JSON.
 3. response.json convierte el JSON a un diccionario de python.

```python
# Manejo de respuestas 

response = requests.get("https://api.github.com/users/octocat")

# 1. Código de estado
print(f"Status: {response.status_code}")  # 200

# 2. Verificar si fue exitoso (200-299)
print(response.ok)  # True si 200-299

# 3. Obtener el contenido como texto
print(response.text[:100])  # Primeros 100 caracteres

# 4. Como JSON (diccionario)
data = response.json()
print(f"Usuario: {data['login']}")
print(f"Repositorios: {data['public_repos']}")

# 5. Headers
print(response.headers['Content-Type'])  # 'application/json; charset=utf-8'
print(response.headers.get('Server'))    # 'GitHub.com'

# 6. Tiempo de respuesta
print(f"Tiempo: {response.elapsed.total_seconds()} segundos")

# 7. URL final (si hubo redirección)
print(response.url)

# 8. Historial de redirecciones
print(response.history)  # Lista de respuestas previas
```

```python
# Parametros en la URL

# Dos formas de pasar parámetros

# 1) Directo en la URL (String)

url = "https://dummyjson.com/products?limit=5&skip=10"
response = requests.get(url)

print(response.url)  # Verás la URL completa
print(response.json()['products'][0]["title"])

# response.json()["products"]
```

```python
# Usando params 

# Definir los parámetros en un diccionario
params = {
    'limit': 5,      # Número de resultados
    'skip': 10,      # Cuántos saltar (para paginación)
    'select': 'title,price,brand'  # Seleccionar campos específicos
}

response = requests.get("https://dummyjson.com/products", params=params)

# La URL se construye automáticamente
print(f"URL construida: {response.url}")  # Verás que todo está codificado

# Obtener los datos y mostrarlos
productos = response.json()['products']
for producto in productos:
    print(f"Producto: {producto['title']}, Precio: {producto['price']}, Marca: {producto['brand']}")
```

Métodos HTTP principales

Con DummyJSON puedes practicar todos los métodos. Recuerda que los cambios con POST, PUT y DELETE son simulados; los datos no se guardan realmente, pero la respuesta que obtienes es idéntica a la de una API real, lo cual es perfecto para aprender .

```python
# Obtener datos GET, la mas común 
response = requests.get("https://dummyjson.com/users")
usuarios = response.json()['users']
print(f"Total de usuarios: {len(usuarios)}")
response.json()
```

```python
# POST crear datos 
# Añade un nuevo producto. Importante: Usa json={...} para que la API sepa que estás enviando datos en formato JSON .

nuevo_producto = {
    'title': 'Mi Producto de Prueba',
    'price': 99.99
}

response = requests.post("https://dummyjson.com/products/add", json=nuevo_producto)

if response.status_code == 201:  # 201 es "Created"
    print("¡Producto creado! (simulado)")
    print(response.json())
else:
    print(f"Error: {response.status_code}")
```

```python
# PUT → Actualizar completamente
# Actualiza un producto existente. Reemplaza el 1 con el ID de otro producto si lo deseas.

producto_actualizado = {
    'title': 'Mi Producto Actualizado',
    'price': 49.99,
    'stock': 100
}

response = requests.put("https://dummyjson.com/products/1", json=producto_actualizado)
print(f"Estado: {response.status_code}")
print(response.json())
```

```python
# DELETE → Eliminar
# Elimina un producto.

response = requests.delete("https://dummyjson.com/products/1")
print(f"Estado: {response.status_code}")
print(response.json())  # Verás que el producto tiene 'isDeleted': true
```

### Autenticación 

```python
# DummyJSON te permite practicar la autenticación con tokens, algo muy común en el mundo real
# Paso 1: Obtener un token de autenticación 

import requests

login_data = {
    'username': 'emilys',  # Usuario de prueba proporcionado por DummyJSON
    'password': 'emilyspass'
}

response = requests.post("https://dummyjson.com/auth/login", json=login_data)

if response.status_code == 200:
    token = response.json()['accessToken']
    print(f"Token obtenido: {token}")
else:
    print("Error al iniciar sesión")
```

```python
# Paso 2:
# Usar el token para hacer peticiones autenticadas 

headers = {
    'Authorization': f'Bearer {token}'  # Es importante el 'Bearer '
}

response = requests.get("https://dummyjson.com/auth/me", headers=headers)
print(response.json())  # Datos del usuario autenticado (emilys)
```

### Manejo de errores 

- Errores 404 (No encontrado): Pide un producto que no existe.

- Errores 400 (Petición incorrecta): Intenta crear un producto sin el title que es requerido.

- Errores de conexión: Simula que tu red falla o que la API no responde con el parámetro delay. DummyJSON te permite añadir un delay para simular una red lenta .


```python
import requests
from requests.exceptions import RequestException, Timeout

url = "https://dummyjson.com/products/9999"  # Producto que no existe

try:
    response = requests.get(url, timeout=2)  # Timeout de 2 segundos
    response.raise_for_status()  # Lanza excepción si el status es 4xx o 5xx
    print(response.json())
    
except requests.exceptions.HTTPError as e:
    print(f"Error HTTP: {e}")
    if response.status_code == 404:
        print("El producto no existe.")
    elif response.status_code == 401:
        print("No estás autenticado.")
    # ... manejar otros códigos de estado

except Timeout:
    print("La petición tardó demasiado en responder.")
    
except RequestException as e:
    print(f"Ocurrió un error: {e}")
```

### Sesiones 

Las sesiones cson útiles cuando tienes que hacer múltiples peticiones a una API que requiere autenticación o un mismo User-Agent. Con DummyJSON, podrías usarla para no tener que ásar el token de autenticación en cada petición.
