```python
# Sexto ejercicio de panda
# Calculate Special Bonus
# Write a solution to calculate the bonus of each employee.
# The bonus of an employee is 100% of their salary if the ID of the employee is an odd number
#  and the employee's name does not start with the character 'M'.
# The bonus of an employee is 0 otherwise.

# Return the result table ordered by employee_id.

import pandas as pd

employees_base = {
    "employee_id": [2, 3, 7, 8, 9],
    "name": ["Meir", "Michael", "Addilyn", "Juan", "Kannon"],
    "salary": [3000, 3800, 7400, 6100, 7700]
}

employees = pd.DataFrame(employees_base)

def calculate_special_bonus(employees: pd.DataFrame) -> pd.DataFrame:
    employees["bonus"] = 0
    employees.loc[(employees["employee_id"]%2 != 0 ) & (employees["name"]\
                                                        .str[0]!="M"),"bonus"] = employees["salary"]
    return employees[["employee_id","bonus"]].sort_values(by="employee_id",ascending=True)

# Una función alternativa 

def calculate_special_bonus1(employees: pd.DataFrame) -> pd.DataFrame:
    from numpy import where
    employees["bonus"] = where(
        (employees["employee_id"]%2 != 0) &
        (employees["name"].str[0] != "M"),
        employees["salary"],
        0
    )
    return employees[["employee_id","bonus"]]

calculate_special_bonus(employees)

```

```python
# # Octavo ejercicio de pandas 
# # Find Users With Valid E-Mails
# # Write a solution to find the users who have valid emails.

# # A valid e-mail has a prefix name and a domain where:

# The prefix name is a string that may contain letters 
# (upper or lower case), digits, underscore '_', period '.', and/or dash '-'. 
# The prefix name must start with a letter.
# The domain must be exactly '@leetcode.com' in lowercase.

import pandas as pd

users_base = {
    "user_id": [1, 2, 3, 4, 5, 6, 7],
    "name": [
        "Winston",
        "Jonathan",
        "Annabelle",
        "Sally",
        "Marwan",
        "David",
        "Shapiro"
    ],
    "mail": [
        "winston@leetcode.com",
        "jonathanisgreat",
        "bella-@leetcode.com",
        "sally.come@leetcode.com",
        "quarz#2020@leetcode.com",
        "david69@gmail.com",
        ".shapiro@leetcode.com"
    ]
}

users = pd.DataFrame(users_base)

def valid_emails(users: pd.DataFrame) -> pd.DataFrame:
    users["col_separado"] = users["mail"].str.split("@")
    users["dominio"] = [x[-1] for x in users["col_separado"]]
    users["pre_nombre"] = [x[0] for x in users["col_separado"]]
    users = users.loc[(users["dominio"] == "leetcode.com") &  (users["pre_nombre"].str.fullmatch(r"[A-Za-z][A-Za-z0-9_.-]*"))]
    return users[["user_id","name","mail"]]

valid_emails(users)
```

```python
# # decimo ejercicio de pandas 
# # Nth Highest Salary
# Write a solution to find the nth highest distinct salary from the Employee table. 
# If there are less than n distinct salaries, return null.

import pandas as pd

employee_base = {
    "id": [1, 2, 3, 4],
    "salary": [100, 200, 300,100]
}

employee = pd.DataFrame(employee_base)

def nth_highest_salary(employee: pd.DataFrame, N: int) -> pd.DataFrame:

    employee = employee.sort_values(by = "salary",ascending=False)
    employee.drop_duplicates(subset=["salary"] ,inplace=True)
    employee["indicador"] = range(1,employee.shape[0]+1)
    employee = employee[employee["indicador"] == N]

    if len(employee) == 0:
        bd = pd.DataFrame({f"getNthHighestSalary({N})":[None]})
        return bd
    else:
        bd = employee.rename(columns={"salary":f"getNthHighestSalary({N})"})
        return pd.DataFrame(bd.iloc[:, 1])


# nth_highest_salary(employee,2)
```

```python
#  Décimo septimo ejercicio de pandas

# Find Total Time Spent by Each Employee
# Write a solution to calculate the total time in minutes spent by each employee on each day at the office.
# Note that within one day, an employee can enter and leave more than once. 
# The time spent in the office for a single entry is out_time - in_time.

import pandas as pd

employees_base = {
    "emp_id": [1, 1, 1, 2, 2],
    "event_day": [
        "2020-11-28",
        "2020-11-28",
        "2020-12-03",
        "2020-11-28",
        "2020-12-09"
    ],
    "in_time": [4, 55, 1, 3, 47],
    "out_time": [32, 200, 42, 33, 74]
}

employees = pd.DataFrame(employees_base)

print(employees)

def total_time(employees: pd.DataFrame) -> pd.DataFrame:

    employees["total_time"] =  employees["out_time"] - employees["in_time"]
    return employees.groupby(["event_day","emp_id"])[["total_time"]].sum().reset_index().\
    rename(columns={"event_day":"day"})

total_time(employees)
    
```

```python
# Vigésimo segundo ejercicio de pandas
# Group Sold Products By The Date

# Write a solution to find for each date the number of different products sold and their names.
# The sold products names for each date should be sorted lexicographically.
# Return the result table ordered by sell_date.

import pandas as pd

activities_base = {
    "sell_date": [
        "2020-05-30",
        "2020-06-01",
        "2020-06-02",
        "2020-05-30",
        "2020-06-01",
        "2020-06-02",
        "2020-05-30"
    ],
    "product": [
        "Headphone",
        "Pencil",
        "Mask",
        "Basketball",
        "Bible",
        "Mask",
        "T-Shirt"
    ]
}

activities = pd.DataFrame(activities_base)

print(activities)

def categorize_products(activities: pd.DataFrame) -> pd.DataFrame:
    activities["sell_date"] = pd.to_datetime(activities["sell_date"])
    return activities.groupby("sell_date")\
        .agg(
        num_sold=("product",lambda x:len(x.unique())),
        products = ("product",lambda x: ",".join(sorted(x.unique())))
        ).reset_index()

categorize_products(activities)
```

```python
# vigésimo sexto ejercicio de pandas 

# Students and Examinations

# Write a solution to find the number of times each student attended each exam.

# Return the result table ordered by student_id and subject_name.

students_base = {
    "student_id": [1, 2, 13, 6],
    "student_name": [None, "Bob", "John", "Alex"]
}

students = pd.DataFrame(students_base)

print(students)


subjects_base = {
    "subject_name": ["Math", "Physics", "Programming"]
}

subjects = pd.DataFrame(subjects_base)

print(subjects)

examinations_base = {
    "student_id": [1, 1, 1, 2, 1, 1, 13, 13, 13, 2, 1],
    "subject_name": [
        "Math",
        "Physics",
        "Programming",
        "Programming",
        "Physics",
        "Math",
        "Math",
        "Programming",
        "Physics",
        "Math",
        "Math"
    ]
}

examinations = pd.DataFrame(examinations_base)

print(examinations)

import pandas as pd

def students_and_examinations(students: pd.DataFrame, subjects: pd.DataFrame, examinations: pd.DataFrame) -> pd.DataFrame:
    bd = pd.merge(students,subjects,how="cross")
    bd1 = pd.merge(bd,examinations,how="left", left_on="student_id",right_on="student_id")
    bd1
    bd1["attended_exams"] = (
        bd1["subject_name_x"] == bd1["subject_name_y"]
    ).astype(int)

    bd1["attended_exams"] = (
        bd1.groupby(["student_id", "subject_name_x"])["attended_exams"]
        .transform("sum")
    )
    bd2 = (bd1.groupby(["student_id","subject_name_x"])
        ["attended_exams"].max().
        reset_index())
    bd3 = pd.merge(bd2,students,how="left",left_on="student_id",right_on="student_id")
    bd3 = bd3.sort_values(by=["student_id","student_name"],ascending=True)\
        .rename(columns={"subject_name_x":"subject_name"})
    return bd3[["student_id","student_name","subject_name","attended_exams"]]

students_and_examinations(students,subjects,examinations)
```

```python
# Vigésimo séptimo ejercicio de pandas 

# Managers with at Least 5 Direct Reports

# Write a solution to find managers with at least five direct reports.

import pandas as pd

employee_base = {
    "id": [101, 102, 103, 104, 105, 106],
    "name": ["John", "Dan", "James", "Amy", "Anne", "Ron"],
    "department": ["A", "A", "A", "A", "A", "B"],
    "managerId": [None, 101, 101, 101, 101, 101]
}

employee = pd.DataFrame(employee_base)

print(employee)

def find_managers(employee: pd.DataFrame) -> pd.DataFrame:
    bd = employee["managerId"].value_counts().reset_index()
    l1 = bd.loc[bd["count"]>=5,["managerId"]]
    l1 = l1["managerId"].astype("int").tolist()
    bdr = employee.loc[employee["id"].isin(l1)][["name"]]
    return bdr

find_managers(employee)
```

```python
# vigésimo octavo ejercicio de pandas

# Sales Person 

# Write a solution to find the names of all the salespersons who did not have any orders 
# related to the company with the name "RED".

import pandas as pd

# 1. Crear la tabla SalesPerson
salesperson_data = {
    'sales_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Amy', 'Mark', 'Pam', 'Alex'],
    'salary': [100000, 12000, 65000, 25000, 5000],
    'commission_rate': [6, 5, 12, 25, 10],
    'hire_date': ['4/1/2006', '5/1/2010', '12/25/2008', '1/1/2005', '2/3/2007']
}
df_salesperson = pd.DataFrame(salesperson_data)

# 2. Crear la tabla Company
company_data = {
    'com_id': [1, 2, 3, 4],
    'name': ['RED', 'ORANGE', 'YELLOW', 'GREEN'],
    'city': ['Boston', 'New York', 'Boston', 'Austin']
}
df_company = pd.DataFrame(company_data)

# 3. Crear la tabla Orders
orders_data = {
    'order_id': [1, 2, 3, 4],
    'order_date': ['1/1/2014', '2/1/2014', '3/1/2014', '4/1/2014'],
    'com_id': [3, 4, 1, 1],
    'sales_id': [4, 5, 1, 4],
    'amount': [10000, 5000, 50000, 25000]
}
df_orders = pd.DataFrame(orders_data)

# --- Imprimir las tablas para verificar ---
print("Tabla SalesPerson:")
print(df_salesperson)
print("\n" + "="*40 + "\n")

print("Tabla Company:")
print(df_company)
print("\n" + "="*40 + "\n")

print("Tabla Orders:")
print(df_orders)

def sales_person(sales_person: pd.DataFrame, company: pd.DataFrame, orders: pd.DataFrame) -> pd.DataFrame:
        bd1 = pd.merge(orders,company, how="left", left_on="com_id",right_on="com_id")
        bd2 = pd.merge(bd1,sales_person, how="left",left_on="sales_id",right_on="sales_id")
        bd3 = bd2.loc[bd2["name_x"] == "RED"]
        l = bd3["name_y"].tolist()

        bd4 = sales_person.loc[~sales_person["name"].isin(l)][["name"]]
        return bd4

sales_person(df_salesperson,df_company,df_orders)
```

```python
# Ejercicios con python - Manejo de archivos
# --------------------------------------------------------------------- #
# Lectura de las bases de datos para el ejercicio

# --------------------------------------------------------------------- #

import pandas as pd

info_tarjetas = pd.read_excel("base_de_datos.xlsx",sheet_name="informacion_tarjetas")
info_cliente = pd.read_excel("base_de_datos.xlsx",sheet_name="informacion_cliente")
transa = pd.read_excel("base_de_datos.xlsx",sheet_name="transaccional")
```

```python
# Pregunta 1:

# Determine cuantas transacciones fueron realizadas despues del 15 de abril de 2026.
# ---------------------------------------------------------- #

# le damos el formato necesario a la variable de fechas

transa["fecha_compra"] = pd.to_datetime(transa["fecha_compra"],format="%d.%m.%Y")
# Filtramos de la fecha 15 de abril del 2026 en adelante 

num_bef_date = len(transa.loc[transa["fecha_compra"] > "2026-03-15"])
# ------------------------------------------------------------#
print(f"Despues del 15 de abril del 2026 se registraron {num_bef_date} transacciones")
```

```python
# Pregunta 2: 

# Identifique lOS clientes que no registraron ninguna transacción durante el mes de junio de 2026.

# ----------------------------------------------------------- #

# Clientes que registraron transacciones en junio 2026

transa["numero_tarjeta"] = transa["numero_tarjeta"].astype("str")
info_cliente["numero_tarjeta"] = info_cliente["numero_tarjeta"].astype("str")

clie_jun = transa\
    .loc[(transa["fecha_compra"].dt.year == 2026) & (transa["fecha_compra"]\
                                                     .dt.month == 6),"numero_tarjeta"].unique().tolist()

# identificamos de la base de clientes que numeros de tarjetas no estan en clie_jun

no_jun = info_cliente.loc[~info_cliente["numero_tarjeta"].isin(clie_jun),"documento"].tolist()
no_jun = [str(x) for x in no_jun]
# ----------------------------------------------------------- #

print(f"Los documentos de clientes que no tuvieron transacciones en Junio fueron {" , ".join(no_jun)}")
```

```python
# Pregunta 3:

# Calcule el costo mensual total que asume el banco por todas las tarjetas 
# clasificadas en la categoría A.

# ------------------------------------------------------------- #
# Tipos de tarjeta categoria A 

tipo_t_A = info_tarjetas.loc[info_tarjetas["categoria_tarjeta"] == "A","tipo_tarjeta"].tolist()

# identificamos todas las tarjetas de este tipo que estan en circulación 

cant_a = info_cliente.loc[info_cliente["tipo_tarjeta"].isin(tipo_t_A)]

# costo total mensual que asume el banco

costo_mensual = cant_a["tipo_tarjeta"].value_counts().reset_index()\
    .merge(info_tarjetas,on="tipo_tarjeta")

costo_mensual["total"] = costo_mensual["count"]*costo_mensual["costo_mensual_unidad"]

# -------------------------------------------------------------- #
print(f"el total de tarjetas de categoria A que estan en circulación es: {sum(costo_mensual["count"])}")
print(f"El costo total que asume el banco es de: {sum(costo_mensual["total"])}")
```

```python
# Pregunta 4:

# El banco obtiene ingresos equivalentes al 1% del valor de cada transacción.

# La rentabilidad deberá calcularse corno:
# Rentabilidad = Ingresos por comisión - Costos mensuales
# Ingresos por comisión sumatoria del valor_compra 1%
# Costos mensuales cantidad de tarjetas en circulación x costo_mensual_unidad

# Objetivo: Determinar cuáles fueron el top 3 de tipos de tarjeta más rentable durante los meses de
# abril y mayo de 2026.

# -------------------------------------------------------------------------------# 
tar_ab_ma = transa\
    .loc[(transa["fecha_compra"].dt.year==2026) &\
          (transa["fecha_compra"].dt.month == 4) | (transa["fecha_compra"].dt.month == 5)]

tar_ab_ma["comisión"] = [x/100 for x in tar_ab_ma["valor_compra"]]

# agregamos el tipo de tarjeta agrupamos y unimos para obtener el costo por unidad 
tar_ab_ma_t = tar_ab_ma\
    .merge(info_cliente[["numero_tarjeta","tipo_tarjeta"]], on="numero_tarjeta")\
    .groupby("tipo_tarjeta")["comisión"].agg(["sum","count"]).reset_index()

cos = tar_ab_ma_t\
    .merge(info_tarjetas[["tipo_tarjeta","costo_mensual_unidad"]],on="tipo_tarjeta")

# Calculamos las columnas necesarias e imprimimos los tres productos mas importantes 
cos["valor_t_t"] = cos["sum"]*100
cos["costo_total"] = cos["count"]*cos["costo_mensual_unidad"]
cos["rentabilida"] = cos["sum"]-cos["costo_total"]
cos.sort_values(by="rentabilida",ascending=False).head(3)
```
