<h3 style="text-align: center;"><strong>Definición y Contexto De Las Variables De La Base De Datos</strong></h3>

**Contexto de la base de datos:**  
Este conjunto de datos contiene anuncios inmobiliarios de EE.UU. desglosados por estado y código postal.

**Variables**

- intermediado por (agencia/intermediario codificado categóricamente)
- estado (Estado de la vivienda: a. lista para la venta o b. lista para construir)
- precio (Precio de la vivienda, ya sea el precio de venta actual o el precio de venta reciente si la casa se vendió recientemente)
- cama (número de camas)
- baño (número de baños)
- acre_lot (Tamaño de la propiedad/terreno en acres)
- calle (dirección postal codificada categóricamente)
- ciudad (nombre de la ciudad)
- estado (nombre del estado)
- código_postal (código postal de la zona)
- tamaño_de_la_casa (superficie/tamaño/espacio habitable de la casa en pies cuadrados)
- prev_sold_date (Fecha de venta anterior)

```python
##---------------------------------------------------##
##------ importación de librerias a usar ------------##

import pandas as pd 
import numpy as np
```

```python
# Lectura de la base de datos 
#--------------------------------------- #  
df = pd.read_csv("realtor-data.zip.csv")

# Encabezado de la base de datos
print(list(df.columns))

df.head()
```


**¿Cual es el precio promedio de las propiedades?** 

```python
Promedio_propiedades = int(df["price"].mean())
print(f"El precio promedio de las propiedades es de {Promedio_propiedades}")
```

**¿Que ciudades tienen las propiedades mas caras?** 

```python
# Definimos los 20 propiedades mas caras y luego obtenemos las ciudades a las quye pertenecen
# --------------------------------------------------------------------------------------------#

ciudades = list(df.sort_values(by=["price"], ascending = False).head(20)["city"].unique())
promedio = int(df.sort_values(by=["price"], ascending = False).head(20)["price"].mean())
print("Las ciudades con las propiedades mas caras son:", ", ".join(ciudades)+f"\nCon un promedio de {promedio}")
```

**¿Cual es el precio promedio por estado?** 

```python
promedio_estado = df.groupby(["state"])["price"].sum()
int(promedio_estado.mean())
print(f"El precio promedio de propiedaes por estado es de: {int(promedio_estado.mean())}")
```

**¿Donde existe mayor cantidad de propiedades?** 

```python
# Ciudad con mayor cantidad de propiedades
pro_c = df.groupby("city")["price"].count().reset_index().sort_values(by=["price"],ascending=False).head(1)
pro_s =  df.groupby("state")["price"].count().reset_index().sort_values(by=["price"],ascending=False).head(1)
print(f"La ciudad con mayor numero de propiedades es : {pro_c.iloc[0,0]} con {pro_c.iloc[0,1]} propiedades" )
print(f"El estado con mayor numero de propiedades es : {pro_s.iloc[0,0]} con {pro_s.iloc[0,1]} propiedades" )
```
