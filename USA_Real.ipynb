<h3 style="text-align: center;"><strong>Definición y Contexto De Las Variables De La Base De Datos</strong></h3>

**Contexto de la base de datos:**  
Este conjunto de datos contiene anuncios inmobiliarios de EE.UU. desglosados por estado y código postal.

**Variables**

- intermediado por (agencia/intermediario codificado categóricamente)
- estado (Estado de la vivienda: a. lista para la venta o b. lista para construir)
- precio (Precio de la vivienda, ya sea el precio de venta actual o el precio de venta reciente si la casa se vendió recientemente)
- cama (número de camas)
- baño (número de baños)
- acre_lot (Tamaño de la propiedad/terreno en acres)
- calle (dirección postal codificada categóricamente)
- ciudad (nombre de la ciudad)
- estado (nombre del estado)
- código_postal (código postal de la zona)
- tamaño_de_la_casa (superficie/tamaño/espacio habitable de la casa en pies cuadrados)
- prev_sold_date (Fecha de venta anterior)
